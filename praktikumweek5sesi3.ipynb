```python
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist, squareform
import matplotlib.pyplot as plt

data = np.array([
    [1.0, 1.0],
    [1.5, 1.5],
    [5.0, 5.0],
    [3.0, 4.0],
    [4.0, 4.0],
    [3.0, 3.5]
])

distance_matrix = pdist(data, metric='euclidean')
distance_df = pd.DataFrame(squareform(distance_matrix),
                           columns=['a', 'b', 'c', 'd', 'e', 'f'],
                           index=['a', 'b', 'c', 'd', 'e', 'f'])

print("=== Distance Matrix (Euclidean) ===")
print(distance_df)

linked = linkage(data, method='single')

plt.figure(figsize=(12, 8))
plt.title("Hierarchical Clustering Dendrogram (Single Linkage)")
dendrogram(linked, labels=['a', 'b', 'c', 'd', 'e', 'f'], leaf_rotation=90)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

data = np.array([
    [1.0, 1.0],
    [1.5, 1.5],
    [5.0, 5.0],
    [3.0, 4.0],
    [4.0, 4.0],
    [3.0, 3.5]
])

labels = ['a', 'b', 'c', 'd', 'e', 'f']
```

```python
linked_single = linkage(data, method='single')
linked_complete = linkage(data, method='complete')
linked_average = linkage(data, method='average')
```

```python
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
dendrogram(linked_single, labels=labels, leaf_rotation=90)
plt.title("Single Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

plt.subplot(1, 3, 2)
dendrogram(linked_complete, labels=labels, leaf_rotation=90)
plt.title("Complete Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

plt.subplot(1, 3, 3)
dendrogram(linked_average, labels=labels, leaf_rotation=90)
plt.title("Average Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

plt.tight_layout()
plt.show()
```

Kode ini dibuat untuk melakukan analisis pengelompokan data atau clustering secara hierarki pada dataset sederhana yang terdiri dari enam titik data. Pada tahap awal, kode menyiapkan titik-titik data dua dimensi beserta labelnya mulai dari huruf a sampai f, lalu memproses data tersebut menggunakan tiga metode penggabungan cluster yang berbeda, yaitu single linkage, complete linkage, dan average linkage. Tujuannya adalah untuk melihat bagaimana perbedaan cara menghitung jarak antar kelompok data dapat memengaruhi struktur pembentukan cluster dari data yang sama.

Metode single linkage bekerja dengan cara mengukur jarak terdekat antara dua anggota kelompok, sehingga sangat peka terhadap titik data yang letaknya berdekatan dan cenderung membentuk pola seperti rantai. Sementara itu, complete linkage menggunakan jarak terjauh antara anggota kelompok sehingga menghasilkan cluster yang lebih padat dan kompak. Di sisi lain, average linkage mengambil rata-rata jarak dari seluruh kombinasi titik data antar kelompok, yang memberikan hasil pertengahan dan lebih seimbang dibandingkan dua metode sebelumnya.

Setelah proses perhitungan linkage selesai, kode memanfaatkan library matplotlib untuk menampilkan hasil visualisasinya berupa diagram pohon yang disebut dendrogram secara berdampingan. Dengan meletakkannya dalam satu tampilan grafik tiga panel, kita bisa langsung membandingkan garis-garis penggabungan data pada masing-masing metode secara visual. Sumbu vertikal pada grafik menunjukkan tingkat jarak atau ketidakmiripan saat dua kelompok mulai digabungkan, sedangkan sumbu horizontal berisi label data yang sedang dianalisis.

Melalui plot perbandingan ini, kita bisa memahami secara praktis bahwa pilihan metode linkage sangat memengaruhi hasil akhir pengelompokan data. Perbedaan tinggi garis pada dendrogram menunjukkan seberapa jauh suatu data harus digabungkan dengan data lainnya. Dengan begitu, kita dapat memilih metode mana yang paling sesuai dengan karakteristik data yang kita miliki, apakah butuh pengelompokan yang ketat, fleksibel, atau yang mengambil nilai rata-rata.
